# Data Cleaning (Python Basics)

**Scenario:**
Imagine you are preparing a dataset of images to feed into an AI model (like a Convolutional Neural Network or CNN). The raw data received from the camera sensor is stored as a list of numbers (pixel values), but unfortunately, the sensor introduced some noise and calculation errors.

**Your Mission:**
In the following cells, you will write a script step-by-step that inspects and cleans this raw data. Fill in the blanks (`___`) and run each cell to apply the cleaning rules!

#### Step 1: Input, Output & Initialization
Ask the user to input the maximum allowed pixel value and convert it to an integer in a *single line* using nested functions. Then, initialize a boolean error flag (mind the capitalization) and create an empty list using Python's built-in function (do not use `[]`).

```python
# Get user input and convert it to an integer simultaneously
max_val = int(input("Enter maximum pixel value (e.g., 255): "))

# Initialize the boolean flag 
has_error = False

raw_pixels = [120, -15, 256, 100, 310, 45, 0, -5, 200]

# Create an empty list using the built-in function
processed_pixels = list()

print(f"Ready to process {len(raw_pixels)} pixels with a max limit of {max_val}.")
```

#### Step 2: Iteration, Flow Control & List Methods
Set up a loop to iterate through the `raw_pixels` list. 
1. If a number is negative, flag the error and bypass the current iteration immediately using the exact flow-control keyword.
2. If the data is out-of-bounds (greater than `max_val`), cap it and add it to your new list using the correct list method.
3. Otherwise, append the valid pixel exactly as it is.

```python
for pixel in raw_pixels:
    
    # Check for negative values (corrupted data)
    if pixel < 0:
        print(f"Warning: Negative value ({pixel}) detected! Skipping...")
        has_error = True
        continue  # Skip the rest of THIS loop iteration
        
    # Check if the pixel value exceeds the maximum limit
    if pixel > max_val: 
        processed_pixels.append(max_val)
        
    # For all other normal values
    else:
        processed_pixels.append(pixel)

print("\nData processing complete!")
```

#### Step 3: Final Output & Identity Operators
Use Python's **identity operator** (do NOT use the equality `==` operator) to check if your boolean flag is EXACTLY the `True` object. Finally, print the cleaned dataset.

```python
# Check if 'has_error' is EXACTLY the True object
if has_error is True: 
    print("Process finished: Corrupted data was found and removed.")
else:
    print("Process finished: All data was originally clean.")

# Print the final result
print("\nProcessed Data:", processed_pixels)
```

# Data Normalization Pipeline (Functions)

**Scenario:**
In machine learning, numerical features are often transformed into a range between 0 and 1. This helps the model train faster and prevents larger numbers from dominating the AI's learning process. 
The basic formula for Min-Max Normalization is: `(current_value - min_value) / (max_value - min_value)`

**Your Mission:**
In the following cells, you will build a highly flexible data processing pipeline step-by-step. Fill in the blanks (`___`) and run each cell to see how higher-order functions and arbitrary arguments (`*args`) work in practice!

#### Step 1: Default Arguments & Returning Values
Let's define a global counter to track our operations. Then, define the `min_max_scale` function. 
Assign a default value of `0` to `min_val` and `100` to `max_val` using the correct assignment operator, and finally, output the calculated value.

```python
transformation_count = 0

def min_max_scale(value, min_val=0, max_val=100):
    
    scaled = (value - min_val) / (max_val - min_val)
    
    return scaled

print("Global counter and scaling function defined successfully!")
```

#### Step 2: Higher-Order Functions, Scope & *args
Now, build the `process_data` manager function. 
1. Pack any extra positional arguments using the correct syntax (`___`).
2. Explicitly declare that you are modifying the GLOBAL `transformation_count` variable.
3. Call the passed function (`transform_func`) on the item and UNPACK the extra arguments into it.
4. Use the compound assignment operator to increment the counter by 1.

```python
def process_data(data_list, transform_func, *args):
    
    global  transformation_count
    
    result_list = []
    
    for item in data_list:
        new_item = transform_func(item, *args)
        result_list.append(new_item)
        
    transformation_count += 1
    
    return result_list

print("Manager function 'process_data' is ready!")
```

#### Step 3: Execution (Passing by Reference)
Now it's time to run the pipeline! Call `process_data` and pass the `min_max_scale` function ITSELF (pass it by reference, without parentheses!). Then print the results.

```python
raw_data = [20, 50, 80, 10]
data_min = 10  # Actual minimum in our dataset
data_max = 80  # Actual maximum in our dataset

normalized_data = process_data(raw_data, min_max_scale, data_min, data_max)

print("Original Data:", raw_data)
print("Normalized Data:", normalized_data)
print("Total Pipeline Executions:", transformation_count)
```

# Building a Model Blueprint (OOP Basics)

**Scenario:**
In modern AI frameworks (like PyTorch or TensorFlow), neural networks and machine learning models are heavily based on Object-Oriented Programming (OOP). You don't just write loose functions; instead, you create a `Class` (a blueprint) for your model, and then generate `Objects` (instances) from that blueprint.

**Your Mission:**
In the following cells, you will define a blueprint for a basic Machine Learning model and then bring it to life. Fill in the blanks (`___`) and run each cell!

#### Step 1: Defining the Class, Constructor, and Methods
1. Use the correct keyword to define a new blueprint named `SimpleModel`.
2. Define the constructor method (remember the double underscores around `init`).
3. Attach the `model_name` and `learning_rate` arguments to the object using the `self` keyword. Do the same for the `is_trained` attribute.
4. Define the `train` method. Remember that ALL instance methods in Python must take a specific first parameter!

```python
class SimpleModel:
    
    def __init__(self, model_name, learning_rate):
        
        self.name = model_name
        self.lr = learning_rate
        self.is_trained = False
        print(f"Model '{self.name}' created with Learning Rate: {self.lr}")

    def train(self, epochs):
        
        print(f"Training '{self.name}' for {epochs} epochs...")
        self.is_trained = True
        print("Training complete!")

print("Blueprint (Class) successfully defined!")
```

#### Step 2: Instantiating and Using the Object
Now that the blueprint is ready, let's build an actual model from it!
1. Instantiate an object from the `SimpleModel` class. Name the model `"VisionNet"` and set the learning rate to `0.001`.
2. Call the `train` method on your newly created object and pass `10` as the number of epochs.

```python
my_ai_model = SimpleModel("VisionNet", 0.001)

my_ai_model.train(10)

# Check the object's attribute to see if its state changed after training
print("Is the model trained?", my_ai_model.is_trained)
```

# Saving Training Logs (File Handling)

**Scenario:**
When training an AI model, the process can take hours or even days. If you don't save the model's performance metrics (like accuracy and loss) to a file, all that information is lost as soon as the program finishes running! 

**Your Mission:**
In the following cells, you will write a list of training logs to a text file on your hard drive, and then read the file back to verify its contents. Fill in the blanks (`___`) and run each cell!

#### Step 1: Writing Logs to a File
Always use the Context Manager (`with`) when opening files. It acts as a safety mechanism that automatically closes the file for you when the block of code finishes. 
Use the built-in function to `open` the file in **Write Mode** (`"w"`) and save the logs line by line using the correct file method.

```python
training_logs = [
    "Epoch 1: Accuracy = 0.75, Loss = 0.50\n",
    "Epoch 2: Accuracy = 0.82, Loss = 0.40\n",
    "Epoch 3: Accuracy = 0.89, Loss = 0.30\n"
]

with open("metrics.txt", "w") as file:
    
    for log in training_logs:
        file.write(log)
        
print("Logs successfully saved to 'metrics.txt'!")
```

#### Step 2: Reading from a File
Now, open the exact same file, but this time in **Read Mode** (`"r"`). Don't forget the `as` keyword to assign it to a variable! 
Then, use the correct file method to read the *entire* content of the file at once and print it.

```python
with open("metrics.txt", "r") as file:
    
    saved_content = file.read()
    
print("--- Displaying Saved Logs ---")
print(saved_content)
```
